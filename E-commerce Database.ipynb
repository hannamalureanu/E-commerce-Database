````python
import pickle
import os
import struct
import re
from typing import List, Tuple, Optional, Dict, Set, Any
from collections import defaultdict

# ========================
# 1. CLASA PRODUCT
# ========================
class Product:
    def __init__(self, id: int, name: str, category: str, price: float, stock: int):
        self.id = id
        self.name = name
        self.category = category
        self.price = price
        self.stock = stock

    def __repr__(self):
        return f"Product(ID={self.id}, Name='{self.name}', Category='{self.category}', Price={self.price}, Stock={self.stock})"

    def __eq__(self, other):
        if not isinstance(other, Product):
            return False
        return (self.id == other.id and self.name == other.name and
                self.category == other.category and self.price == other.price and
                self.stock == other.stock)

    def __hash__(self):
        return hash((self.id, self.name, self.category, self.price, self.stock))

    def to_bytes(self) -> bytes:
        return pickle.dumps(self)

    @staticmethod
    def from_bytes(data: bytes) -> 'Product':
        return pickle.loads(data)


# ========================
# 2. B-TREE - PRIMARY INDEX (ID)
# ========================
class BTreeNode:
    def __init__(self, t: int, leaf: bool):
        self.t = t
        self.leaf = leaf
        self.keys: List[Tuple[int, int]] = []  # (ID, offset)
        self.children: List['BTreeNode'] = []


class BTree:
    def __init__(self, t: int = 3):
        self.t = t
        self.root = BTreeNode(t, True)
        self.size = 0

    def search(self, key: int) -> Optional[int]:
        node = self.root
        while True:
            i = 0
            while i < len(node.keys) and key > node.keys[i][0]:
                i += 1

            if i < len(node.keys) and node.keys[i][0] == key:
                return node.keys[i][1]

            if node.leaf:
                return None
            node = node.children[i]

    def insert(self, key: int, offset: int):
        self.size += 1
        root = self.root

        if len(root.keys) == 2 * self.t - 1:
            new_root = BTreeNode(self.t, False)
            new_root.children.append(self.root)
            self._split_child(new_root, 0)
            self.root = new_root

        self._insert_non_full(self.root, key, offset)

    def _insert_non_full(self, node: BTreeNode, key: int, offset: int):
        i = len(node.keys) - 1

        if node.leaf:
            node.keys.append((0, 0))
            while i >= 0 and key < node.keys[i][0]:
                node.keys[i + 1] = node.keys[i]
                i -= 1
            node.keys[i + 1] = (key, offset)
        else:
            while i >= 0 and key < node.keys[i][0]:
                i -= 1
            i += 1

            if len(node.children[i].keys) == 2 * self.t - 1:
                self._split_child(node, i)
                if key > node.keys[i][0]:
                    i += 1
            self._insert_non_full(node.children[i], key, offset)

    def _split_child(self, parent: BTreeNode, i: int):
        t = self.t
        y = parent.children[i]
        z = BTreeNode(t, y.leaf)

        parent.keys.insert(i, y.keys[t - 1])
        parent.children.insert(i + 1, z)

        z.keys = y.keys[t:]
        y.keys = y.keys[:t - 1]

        if not y.leaf:
            z.children = y.children[t:]
            y.children = y.children[:t]

    def delete(self, key: int) -> bool:
        if self.search(key) is None:
            return False

        self._delete(self.root, key)
        self.size -= 1

        if not self.root.leaf and len(self.root.keys) == 0:
            self.root = self.root.children[0]
        return True

    def _delete(self, node: BTreeNode, key: int):
        idx = 0
        while idx < len(node.keys) and key > node.keys[idx][0]:
            idx += 1

        if idx < len(node.keys) and node.keys[idx][0] == key:
            if node.leaf:
                node.keys.pop(idx)
            else:
                self._delete_internal(node, idx)
        else:
            if node.leaf:
                return

            if len(node.children[idx].keys) < self.t:
                self._fill(node, idx)

            if idx > len(node.keys):
                self._delete(node.children[idx - 1], key)
            else:
                self._delete(node.children[idx], key)

    def _delete_internal(self, node: BTreeNode, idx: int):
        t = self.t

        if len(node.children[idx].keys) >= t:
            predecessor = self._get_predecessor(node.children[idx])
            node.keys[idx] = predecessor
            self._delete(node.children[idx], predecessor[0])
        elif len(node.children[idx + 1].keys) >= t:
            successor = self._get_successor(node.children[idx + 1])
            node.keys[idx] = successor
            self._delete(node.children[idx + 1], successor[0])
        else:
            self._merge(node, idx)
            self._delete(node.children[idx], node.keys[idx][0])

    def _get_predecessor(self, node: BTreeNode) -> Tuple[int, int]:
        while not node.leaf:
            node = node.children[-1]
        return node.keys[-1]

    def _get_successor(self, node: BTreeNode) -> Tuple[int, int]:
        while not node.leaf:
            node = node.children[0]
        return node.keys[0]

    def _fill(self, node: BTreeNode, idx: int):
        t = self.t

        if idx != 0 and len(node.children[idx - 1].keys) >= t:
            self._borrow_from_left(node, idx)
        elif idx != len(node.keys) and len(node.children[idx + 1].keys) >= t:
            self._borrow_from_right(node, idx)
        else:
            if idx != len(node.keys):
                self._merge(node, idx)
            else:
                self._merge(node, idx - 1)

    def _borrow_from_left(self, node: BTreeNode, idx: int):
        child = node.children[idx]
        sibling = node.children[idx - 1]

        child.keys.insert(0, node.keys[idx - 1])
        if not sibling.leaf:
            child.children.insert(0, sibling.children.pop())

        node.keys[idx - 1] = sibling.keys.pop()

    def _borrow_from_right(self, node: BTreeNode, idx: int):
        child = node.children[idx]
        sibling = node.children[idx + 1]

        child.keys.append(node.keys[idx])
        if not sibling.leaf:
            child.children.append(sibling.children.pop(0))

        node.keys[idx] = sibling.keys.pop(0)

    def _merge(self, node: BTreeNode, idx: int):
        child = node.children[idx]
        sibling = node.children[idx + 1]

        child.keys.append(node.keys.pop(idx))
        child.keys.extend(sibling.keys)

        if not child.leaf:
            child.children.extend(sibling.children)
        node.children.pop(idx + 1)

    def range_search(self, start_id: int, end_id: int) -> List[int]:
        results = []
        self._range_traverse(self.root, start_id, end_id, results)
        return results

    def _range_traverse(self, node: BTreeNode, start_id: int, end_id: int, results: List[int]):
        i = 0
        while i < len(node.keys) and node.keys[i][0] < start_id:
            i += 1

        while i < len(node.keys) and node.keys[i][0] <= end_id:
            if node.leaf:
                results.append(node.keys[i][1])
                i += 1
            else:
                self._range_traverse(node.children[i], start_id, end_id, results)
                results.append(node.keys[i][1])
                i += 1

        if not node.leaf and i < len(node.children):
            self._range_traverse(node.children[i], start_id, end_id, results)

    def get_all_offsets(self) -> List[int]:
        offsets = []
        self._traverse(self.root, offsets)
        return offsets

    def _traverse(self, node: BTreeNode, offsets: List[int]):
        for _, offset in node.keys:
            offsets.append(offset)

        if not node.leaf:
            for child in node.children:
                self._traverse(child, offsets)


# ========================
# 3. B+ TREE - SECONDARY INDEX (CATEGORY & PRICE)
# ========================
class BPlusTreeNode:
    def __init__(self, t: int, leaf: bool):
        self.t = t
        self.leaf = leaf
        self.keys: List[Tuple[str, float]] = []  # (category, price)
        self.children: List['BPlusTreeNode'] = []
        self.values: List[List[int]] = []  # list of offsets for each key
        self.next: Optional['BPlusTreeNode'] = None
        self.parent: Optional['BPlusTreeNode'] = None

    def is_full(self) -> bool:
        return len(self.keys) >= 2 * self.t - 1

    def is_minimal(self) -> bool:
        # Pentru rădăcină, minimum este 1 (dacă nu e frunză), altfel 0
        return len(self.keys) < (self.t - 1 if self.parent is not None else 1)

    def __repr__(self):
        if self.leaf:
            return f"LeafNode(keys={self.keys}, values={[len(v) for v in self.values]})"
        else:
            return f"InternalNode(keys={self.keys}, children={len(self.children)})"


class BPlusTree:
    def __init__(self, t: int = 3):
        self.t = t
        self.root = BPlusTreeNode(t, True)
        self.size = 0

    @staticmethod
    def _float_eq(a: float, b: float) -> bool:
        """Compare floats with tolerance"""
        return abs(a - b) < 0.001

    def _find_key_index(self, keys: List[Tuple[str, float]], key: Tuple[str, float]) -> Optional[int]:
        """Find index of key with floating point tolerance"""
        for i, (cat, price) in enumerate(keys):
            if cat == key[0] and self._float_eq(price, key[1]):
                return i
        return None

    def search(self, key: Tuple[str, float]) -> List[int]:
        """Search for exact key (category, price)"""
        node = self.root
        while not node.leaf:
            i = 0
            while i < len(node.keys) and key > node.keys[i]:
                i += 1
            node = node.children[i]

        idx = self._find_key_index(node.keys, key)
        if idx is not None:
            return node.values[idx]
        return []

    def search_range(self, category: str, min_price: float, max_price: float) -> List[int]:
        """Range search for a specific category and price range"""
        results = []
        node = self.root

        # Find first leaf with category >= target
        while not node.leaf:
            i = 0
            while i < len(node.keys) and (category, min_price) > node.keys[i]:
                i += 1
            node = node.children[i]

        # Traverse leaves
        while node:
            for i, (cat, price) in enumerate(node.keys):
                if cat == category and min_price <= price <= max_price:
                    results.extend(node.values[i])
                elif cat > category or (cat == category and price > max_price):
                    break

            node = node.next
            if node and node.keys and node.keys[0][0] > category:
                break

        return results

    def search_by_category(self, category: str) -> List[int]:
        """Get all offsets for a specific category"""
        results = []
        node = self.root

        # Find first leaf with category >= target
        while not node.leaf:
            i = 0
            while i < len(node.keys) and category > node.keys[i][0]:
                i += 1
            node = node.children[i]

        # Traverse all leaves with this category
        while node:
            for i, (cat, _) in enumerate(node.keys):
                if cat == category:
                    results.extend(node.values[i])
                elif cat > category:
                    break

            node = node.next
            if node and node.keys and node.keys[0][0] > category:
                break

        return results

    def insert(self, key: Tuple[str, float], offset: int):
        """Insert a new key-value pair"""
        self.size += 1

        # Inserare normală
        leaf = self._find_leaf(key)
        if leaf is None:
            return

        # Verificăm dacă cheia există deja
        idx = self._find_key_index(leaf.keys, key)
        if idx is not None:
            if offset not in leaf.values[idx]:
                leaf.values[idx].append(offset)
            return

        # Inserăm în frunză
        self._insert_into_leaf(leaf, key, offset)

        # Dacă frunza este prea plină, o împărțim
        if len(leaf.keys) > 2 * self.t - 1:
            self._split_leaf(leaf)

    def _find_leaf(self, key: Tuple[str, float]) -> BPlusTreeNode:
        """Find the leaf node where this key should be inserted"""
        node = self.root
        while not node.leaf:
            i = 0
            while i < len(node.keys) and key > node.keys[i]:
                i += 1
            node = node.children[i]
        return node

    def _insert_into_leaf(self, leaf: BPlusTreeNode, key: Tuple[str, float], offset: int):
        """Insert key into leaf node"""
        # Găsim poziția corectă
        i = 0
        while i < len(leaf.keys) and key > leaf.keys[i]:
            i += 1

        leaf.keys.insert(i, key)
        leaf.values.insert(i, [offset])

    def _split_leaf(self, leaf: BPlusTreeNode):
        """Split a leaf node that is too full"""
        t = self.t
        new_leaf = BPlusTreeNode(t, True)

        # Mutăm jumătate din chei în noua frunză
        mid = len(leaf.keys) // 2
        new_leaf.keys = leaf.keys[mid:]
        new_leaf.values = leaf.values[mid:]
        leaf.keys = leaf.keys[:mid]
        leaf.values = leaf.values[:mid]

        # Actualizăm pointerii next
        new_leaf.next = leaf.next
        leaf.next = new_leaf
        new_leaf.parent = leaf.parent

        # Introducem cheia de separare în părinte
        self._insert_into_parent(leaf, new_leaf.keys[0], new_leaf)

    def _insert_into_parent(self, left: BPlusTreeNode, key: Tuple[str, float], right: BPlusTreeNode):
        """Insert a key and right child into the parent node"""
        if left.parent is None:
            # Creăm o nouă rădăcină
            new_root = BPlusTreeNode(self.t, False)
            new_root.keys = [key]
            new_root.children = [left, right]
            left.parent = new_root
            right.parent = new_root
            self.root = new_root
            return

        parent = left.parent
        # Găsim poziția unde să inserăm cheia
        i = 0
        while i < len(parent.keys) and key > parent.keys[i]:
            i += 1

        # Inserăm cheia și copilul
        parent.keys.insert(i, key)
        parent.children.insert(i + 1, right)
        right.parent = parent

        # Dacă părintele este prea plin, îl împărțim
        if len(parent.keys) > 2 * self.t - 1:
            self._split_internal(parent)

    def _split_internal(self, node: BPlusTreeNode):
        """Split an internal node that is too full"""
        t = self.t
        new_node = BPlusTreeNode(t, False)

        # Cheia de la mijloc va urca
        mid = t - 1
        middle_key = node.keys[mid]

        # Mutăm chei și copii
        new_node.keys = node.keys[mid + 1:]
        new_node.children = node.children[mid + 1:]
        node.keys = node.keys[:mid]
        node.children = node.children[:mid + 1]

        # Actualizăm părinții copiilor
        for child in new_node.children:
            child.parent = new_node

        # Inserăm în părinte
        self._insert_into_parent(node, middle_key, new_node)

    def delete(self, key: Tuple[str, float], offset: int) -> bool:
        """Delete a specific offset for a key"""
        # Găsim frunza
        leaf = self._find_leaf(key)
        if leaf is None:
            return False

        idx = self._find_key_index(leaf.keys, key)
        if idx is None:
            return False

        # Eliminăm offset-ul
        if offset in leaf.values[idx]:
            leaf.values[idx].remove(offset)
            self.size -= 1
        else:
            return False

        # Dacă lista de offset-uri este goală, eliminăm cheia
        if not leaf.values[idx]:
            leaf.keys.pop(idx)
            leaf.values.pop(idx)

            # Rebalansăm dacă este necesar
            if leaf != self.root and len(leaf.keys) < self.t - 1:
                self._rebalance_leaf(leaf)

            # Actualizăm cheile din nodurile interne
            self._update_internal_keys(key, leaf)

        return True

    def _rebalance_leaf(self, leaf: BPlusTreeNode):
        """Rebalance a leaf node that has too few keys"""
        parent = leaf.parent
        if parent is None:
            return

        # Găsim index-ul frunzei în părinte
        try:
            leaf_idx = parent.children.index(leaf)
        except ValueError:
            return

        # Încercăm să împrumutăm de la fratele din stânga
        if leaf_idx > 0:
            left_sibling = parent.children[leaf_idx - 1]
            if len(left_sibling.keys) > self.t - 1:
                # Împrumută de la stânga
                borrowed_key = left_sibling.keys.pop()
                borrowed_values = left_sibling.values.pop()
                leaf.keys.insert(0, borrowed_key)
                leaf.values.insert(0, borrowed_values)
                parent.keys[leaf_idx - 1] = leaf.keys[0]
                return

        # Încercăm să împrumutăm de la fratele din dreapta
        if leaf_idx < len(parent.children) - 1:
            right_sibling = parent.children[leaf_idx + 1]
            if len(right_sibling.keys) > self.t - 1:
                # Împrumută de la dreapta
                borrowed_key = right_sibling.keys.pop(0)
                borrowed_values = right_sibling.values.pop(0)
                leaf.keys.append(borrowed_key)
                leaf.values.append(borrowed_values)
                parent.keys[leaf_idx] = right_sibling.keys[0]
                return

        # Dacă nu putem împrumuta, fuzionăm
        if leaf_idx > 0:
            # Fuziune cu stânga
            left_sibling = parent.children[leaf_idx - 1]
            left_sibling.keys.extend(leaf.keys)
            left_sibling.values.extend(leaf.values)
            left_sibling.next = leaf.next

            # Elimină cheia din părinte
            parent.keys.pop(leaf_idx - 1)
            parent.children.pop(leaf_idx)

            # Rebalansează părintele dacă este necesar
            if parent != self.root and len(parent.keys) < self.t - 1:
                self._rebalance_internal(parent)
            elif parent == self.root and len(parent.keys) == 0:
                # Părintele este rădăcină goală
                self.root = left_sibling
                left_sibling.parent = None

        elif leaf_idx < len(parent.children) - 1:
            # Fuziune cu dreapta
            right_sibling = parent.children[leaf_idx + 1]
            leaf.keys.extend(right_sibling.keys)
            leaf.values.extend(right_sibling.values)
            leaf.next = right_sibling.next

            # Elimină cheia din părinte
            parent.keys.pop(leaf_idx)
            parent.children.pop(leaf_idx + 1)

            # Rebalansează părintele dacă este necesar
            if parent != self.root and len(parent.keys) < self.t - 1:
                self._rebalance_internal(parent)
            elif parent == self.root and len(parent.keys) == 0:
                # Părintele este rădăcină goală
                self.root = leaf
                leaf.parent = None

    def _rebalance_internal(self, node: BPlusTreeNode):
        """Rebalance an internal node that has too few keys"""
        parent = node.parent
        if parent is None:
            return

        try:
            node_idx = parent.children.index(node)
        except ValueError:
            return

        # Încercă împrumut de la stânga
        if node_idx > 0:
            left_sibling = parent.children[node_idx - 1]
            if len(left_sibling.keys) > self.t - 1:
                # Împrumută cheie de la părinte
                parent_key = parent.keys[node_idx - 1]
                borrowed_key = left_sibling.keys.pop()
                borrowed_child = left_sibling.children.pop()

                # Mută cheia din părinte în nod
                node.keys.insert(0, parent_key)
                node.children.insert(0, borrowed_child)
                borrowed_child.parent = node

                # Mută cheia împrumutată în părinte
                parent.keys[node_idx - 1] = borrowed_key
                return

        # Încercă împrumut de la dreapta
        if node_idx < len(parent.children) - 1:
            right_sibling = parent.children[node_idx + 1]
            if len(right_sibling.keys) > self.t - 1:
                # Împrumută cheie de la părinte
                parent_key = parent.keys[node_idx]
                borrowed_key = right_sibling.keys.pop(0)
                borrowed_child = right_sibling.children.pop(0)

                # Mută cheia din părinte în nod
                node.keys.append(parent_key)
                node.children.append(borrowed_child)
                borrowed_child.parent = node

                # Mută cheia împrumutată în părinte
                parent.keys[node_idx] = borrowed_key
                return

        # Fuziune
        if node_idx > 0:
            # Fuziune cu stânga
            left_sibling = parent.children[node_idx - 1]
            parent_key = parent.keys.pop(node_idx - 1)

            # Mută cheia din părinte
            left_sibling.keys.append(parent_key)
            left_sibling.keys.extend(node.keys)
            left_sibling.children.extend(node.children)

            # Actualizează părinții
            for child in node.children:
                child.parent = left_sibling

            parent.children.pop(node_idx)

            # Rebalansează părintele dacă este necesar
            if parent != self.root and len(parent.keys) < self.t - 1:
                self._rebalance_internal(parent)
            elif parent == self.root and len(parent.keys) == 0:
                self.root = left_sibling
                left_sibling.parent = None

        elif node_idx < len(parent.children) - 1:
            # Fuziune cu dreapta
            right_sibling = parent.children[node_idx + 1]
            parent_key = parent.keys.pop(node_idx)

            # Mută cheia din părinte
            node.keys.append(parent_key)
            node.keys.extend(right_sibling.keys)
            node.children.extend(right_sibling.children)

            # Actualizează părinții
            for child in right_sibling.children:
                child.parent = node

            parent.children.pop(node_idx + 1)

            # Rebalansează părintele dacă este necesar
            if parent != self.root and len(parent.keys) < self.t - 1:
                self._rebalance_internal(parent)
            elif parent == self.root and len(parent.keys) == 0:
                self.root = node
                node.parent = None

    def _update_internal_keys(self, deleted_key: Tuple[str, float], leaf: BPlusTreeNode):
        """Update keys in internal nodes after deletion"""
        if not leaf.keys:
            return

        new_smallest = leaf.keys[0]

        # Parcurgem în sus pentru a actualiza cheile
        node = leaf
        while node.parent:
            parent = node.parent
            try:
                child_idx = parent.children.index(node)
            except ValueError:
                break

            if child_idx > 0 and parent.keys[child_idx - 1] == deleted_key:
                parent.keys[child_idx - 1] = new_smallest
                break

            node = parent

    def get_all_offsets(self) -> List[int]:
        """Get all offsets from the B+Tree"""
        results = []
        node = self.root

        while not node.leaf:
            node = node.children[0]

        while node:
            for values in node.values:
                results.extend(values)
            node = node.next
        return results

    def count_keys(self) -> int:
        """Count actual keys in leaves"""
        count = 0
        node = self.root

        while not node.leaf:
            node = node.children[0]

        while node:
            count += len(node.keys)
            node = node.next

        return count

    def count_offsets(self) -> int:
        """Count total offsets stored in leaves"""
        count = 0
        node = self.root

        while not node.leaf:
            node = node.children[0]

        while node:
            for values in node.values:
                count += len(values)
            node = node.next

        return count

    def validate(self) -> Dict[str, Any]:
        """Validate B+tree structure"""
        stats = {
            'valid': True,
            'total_keys': 0,
            'total_offsets': 0,
            'leaf_count': 0,
            'max_depth': 0,
            'errors': []
        }

        if not self.root:
            stats['valid'] = False
            stats['errors'].append("Root is None")
            return stats

        # Check root properties
        if self.root.leaf and len(self.root.keys) == 0:
            stats['valid'] = False
            stats['errors'].append("Root leaf has no keys")

        # Traverse tree
        stack = [(self.root, 0)]
        while stack:
            node, depth = stack.pop()
            stats['max_depth'] = max(stats['max_depth'], depth)

            if node.leaf:
                stats['leaf_count'] += 1
                stats['total_keys'] += len(node.keys)
                for values in node.values:
                    stats['total_offsets'] += len(values)

                # Check leaf order
                for i in range(len(node.keys) - 1):
                    if node.keys[i] >= node.keys[i + 1]:
                        stats['valid'] = False
                        stats['errors'].append(f"Leaf keys not in order: {node.keys[i]} >= {node.keys[i + 1]}")

                # Check that all value lists are non-empty
                for i, values in enumerate(node.values):
                    if not values:
                        stats['valid'] = False
                        stats['errors'].append(f"Empty value list for key {node.keys[i]}")
            else:
                # Check internal node
                if len(node.children) != len(node.keys) + 1:
                    stats['valid'] = False
                    stats['errors'].append(f"Internal node has {len(node.keys)} keys but {len(node.children)} children")

                # Check key order
                for i in range(len(node.keys) - 1):
                    if node.keys[i] >= node.keys[i + 1]:
                        stats['valid'] = False
                        stats['errors'].append(f"Internal keys not in order: {node.keys[i]} >= {node.keys[i + 1]}")

                # Add children to stack
                for child in node.children:
                    if child.parent != node:
                        stats['valid'] = False
                        stats['errors'].append("Parent pointer incorrect")
                    stack.append((child, depth + 1))

        # Check leaf linkage
        node = self.root
        while not node.leaf:
            node = node.children[0]

        leaf_count = 0
        prev = None
        while node:
            leaf_count += 1
            if prev and prev.next != node:
                stats['valid'] = False
                stats['errors'].append("Leaf next pointer broken")
            prev = node
            node = node.next

        if leaf_count != stats['leaf_count']:
            stats['valid'] = False
            stats['errors'].append(f"Leaf count mismatch: traversal={leaf_count}, count={stats['leaf_count']}")

        return stats


# ========================
# 4. PRODUCT DATABASE WITH MULTI-LEVEL INDEX
# ========================
class ProductDatabase:
    def __init__(self, data_file: str = "products.dat"):
        self.data_file = data_file
        self.primary_index = BTree(3)
        self.secondary_index = BPlusTree(3)
        self.deleted_offsets: Set[int] = set()
        self.free_list: List[int] = []

        if os.path.exists(data_file):
            self._load_database()
        else:
            with open(data_file, 'wb') as f:
                pass

    def _write_product(self, product: Product) -> int:
        """Scrie un produs in fisier si returneaza offsetul (pozitia produsului in fisier)"""
        if self.free_list:  #Daca exista loc liber il reutilizeaza
            offset = self.free_list.pop()
            with open(self.data_file, 'r+b') as f:
                f.seek(offset)   #Muta cursorul la pozitia libera din fisier
                data = product.to_bytes()  #Transformă produsul în bytes
                size = len(data)
                f.write(struct.pack('I', size))
                f.write(data)
            if offset in self.deleted_offsets:
                self.deleted_offsets.remove(offset)
            return offset

        with open(self.data_file, 'ab') as f:   #Daca nu exista spatiu liber, scrie produsul la final
            offset = f.tell()   #Ia poziția curentă (sfârșitul fișierului)
            data = product.to_bytes()
            size = len(data)
            f.write(struct.pack('I', size))
            f.write(data)
            return offset

    def _read_product(self, offset: int) -> Optional[Product]:
        """Citeste un produs din fisier folosind offsetul"""
        if offset in self.deleted_offsets:
            return None

        try:
            with open(self.data_file, 'rb') as f:
                f.seek(offset)
                size_data = f.read(4)   #Citește primii 4 bytes
                if len(size_data) < 4:
                    return None
                size = struct.unpack('I', size_data)[0]  #Transformă cei 4 bytes într-un număr

                product_data = f.read(size)  #Citeste datele produsului
                if len(product_data) < size:
                    return None

                return Product.from_bytes(product_data)  #reconstruieste produsul
        except:
            return None

    def _load_database(self):
        """Incarca baza de date din fisier si reconstruieste indexurile B-Tree si B+Tree la pornirea aplicatiei"""
        if not os.path.exists(self.data_file):
            return

        with open(self.data_file, 'rb') as f:
            offset = 0   #Citirea incepe de la inceputul fisierului
            while True:  #Citeste toate produsele din fisier
                f.seek(offset)
                size_data = f.read(4)
                if not size_data:
                    break

                try:
                    size = struct.unpack('I', size_data)[0]
                except:
                    break

                if offset in self.deleted_offsets:  #Daca produsul de la un offset apare ca fiind sters adaugam offsetul la lista de spatii libere
                    self.free_list.append(offset)
                    offset += 4 + size
                    continue

                product_data = f.read(size)
                if len(product_data) < size:
                    break

                try:
                    product = Product.from_bytes(product_data)   #Reconstruieste produsul
                    self.primary_index.insert(product.id, offset)   #Il adauga la B-Tree
                    self.secondary_index.insert((product.category, product.price), offset)  #Il adauga la B+Tree
                except:
                    self.deleted_offsets.add(offset)
                    self.free_list.append(offset)

                offset += 4 + size   #Trece la urmatorul produs

    def insert(self, *args) -> bool:
        """Insereaza produse in baza de date"""
        if len(args) == 1 and isinstance(args[0], Product):  #Daca primeste direct produsul, il insereaza
            product = args[0]
        elif len(args) == 5:   #Daca primeste 5 argumente, creeaza un produs
            product = Product(*args)
        else:
            raise ValueError("Invalid arguments for insert")

        if self.primary_index.search(product.id) is not None:   #Verifica daca ID-ul este unic
            print(f"Error: Product with ID {product.id} already exists")
            return False

        offset = self._write_product(product)
        self.primary_index.insert(product.id, offset)
        self.secondary_index.insert((product.category, product.price), offset)

        print(f"Inserted: {product}")
        return True

    def select(self, condition: str) -> List[Product]:
        """Operatie de tip select cu o conditie WHERE"""
        if ' OR ' in condition.upper():   #Daca exista OR in conditie
            parts = condition.split(' OR ')  #Imparte conditia in doua
            results = set()  #evita dubluri
            for part in parts:   #Parcurge fiecare conditie separat
                part_results = self._process_single_condition(part.strip())  #Execută fiecare condiție simplă
                for product in part_results:
                    results.add(product.id)

            products = []
            for product_id in results:   #Ptr fiecare produs gasit, il returneaza
                offset = self.primary_index.search(product_id)
                if offset is not None:
                    product = self._read_product(offset)
                    if product:
                        products.append(product)
            return products

        return self._process_single_condition(condition.strip())

    def _process_single_condition(self, condition: str) -> List[Product]:
        """Proceseaza o singura conditie WHERE"""
        condition = condition.strip()  #Elimina spatiile inutile din text
        products = []

        # ID = value
        id_match = re.match(r'ID\s*=\s*(\d+)', condition, re.IGNORECASE)
        if id_match:  #Extrage valoarea ID-ului, il cauta in B-Tree si il afisaza
            product_id = int(id_match.group(1))
            offset = self.primary_index.search(product_id)
            if offset is not None:
                product = self._read_product(offset)
                if product:
                    products.append(product)
            return products

        # ID BETWEEN start AND end
        id_range_match = re.match(r'ID\s+BETWEEN\s+(\d+)\s+AND\s+(\d+)', condition, re.IGNORECASE)
        if id_range_match:  #Foloseste range search in B-Tree
            start_id = int(id_range_match.group(1))
            end_id = int(id_range_match.group(2))
            offsets = self.primary_index.range_search(start_id, end_id)
            for offset in offsets:
                product = self._read_product(offset)
                if product:
                    products.append(product)
            return products

        # CATEGORY = 'value'
        cat_match = re.search(r"CATEGORY\s*=\s*'([^']+)'", condition, re.IGNORECASE)
        if cat_match and 'PRICE BETWEEN' not in condition.upper():  #Foloseste B+Tree
            category = cat_match.group(1)
            offsets = self.secondary_index.search_by_category(category)
            for offset in offsets:
                product = self._read_product(offset)
                if product:
                    products.append(product)
            return products

        # CATEGORY = 'value' AND PRICE BETWEEN min AND max
        complex_match = re.search(
            r"CATEGORY\s*=\s*'([^']+)'\s+AND\s+PRICE\s+BETWEEN\s+([\d.]+)\s+AND\s+([\d.]+)",
            condition, re.IGNORECASE
        )
        if complex_match:  #Foloseste B+Tree
            category = complex_match.group(1)
            min_price = float(complex_match.group(2))
            max_price = float(complex_match.group(3))
            offsets = self.secondary_index.search_range(category, min_price, max_price)
            for offset in offsets:
                product = self._read_product(offset)
                if product:
                    products.append(product)
            return products

        # PRICE BETWEEN min AND max
        price_match = re.search(r'PRICE\s+BETWEEN\s+([\d.]+)\s+AND\s+([\d.]+)', condition, re.IGNORECASE)
        if price_match:  #Filtreaza manual produsele dupa pret
            min_price = float(price_match.group(1))
            max_price = float(price_match.group(2))
            all_offsets = self.secondary_index.get_all_offsets()
            for offset in all_offsets:
                product = self._read_product(offset)
                if product and min_price <= product.price <= max_price:
                    products.append(product)
            return products

        return products

    def delete(self, condition: str) -> int:
        """Sterge produsele care respecta conditia WHERE"""
        products_to_delete = self.select(condition)
        deleted_count = 0  #contor ptr nr. produselor sterse

        for product in products_to_delete: #Parcurge fiecare produs
            offset = self.primary_index.search(product.id)
            if offset is not None:
                # Sterge produsul din B-Tree
                if self.primary_index.delete(product.id):
                    # Sterge produsul din B+Tree
                    key = (product.category, product.price)
                    self.secondary_index.delete(key, offset)

                    # Marcheaza produsul ca fiind sters si creste nr. de spatii libere
                    self.deleted_offsets.add(offset)
                    self.free_list.append(offset)
                    deleted_count += 1
                    print(f"Deleted: {product}")

        return deleted_count

    def compact(self):
        """Compacteaza baza de date"""
        temp_file = self.data_file + ".tmp"   #fisier temporar
        new_primary = BTree(3)    #B-Tree gol
        new_secondary = BPlusTree(3)   #B+Tree gol

        with open(temp_file, 'wb') as tf:
            all_products = self.get_all_products()   #Ia toate produsele, mai putin cele sterse

            for product in all_products:
                tf.seek(0, 2)  #Mută cursorul la sfârșitul fișierului temporar
                new_offset = tf.tell()  #Salveaza noua pozitie a produsului

                data = product.to_bytes()
                size = len(data)
                tf.write(struct.pack('I', size))
                tf.write(data)

                #Adauga produsul in noul B-Tree si B+Tree
                new_primary.insert(product.id, new_offset)
                new_secondary.insert((product.category, product.price), new_offset)

        os.replace(temp_file, self.data_file)  #Inlocuieste fisierul vechi cu cel temporar

        self.primary_index = new_primary
        self.secondary_index = new_secondary
        self.deleted_offsets.clear()
        self.free_list.clear()

        print(f"Database compacted. Total products: {new_primary.size}")

    def get_all_products(self) -> List[Product]:
        """Returneaza toate produsele din baza de date"""
        products = []
        all_offsets = self.primary_index.get_all_offsets()

        for offset in all_offsets:
            if offset not in self.deleted_offsets:
                product = self._read_product(offset)
                if product:
                    products.append(product)

        return products

    def statistics(self):
        """Statisticile bazei de date"""
        all_products = self.get_all_products()

        print("\n" + "=" * 80)
        print("DATABASE STATISTICS")
        print("=" * 80)
        print(f"Total products: {len(all_products)}")
        print(f"Deleted offsets: {len(self.deleted_offsets)}")
        print(f"Free list size: {len(self.free_list)}")
        print(f"B-tree (primary) size: {self.primary_index.size}")

        # Use actual counts from B+tree for accurate statistics
        bplus_key_count = self.secondary_index.count_keys()
        bplus_offset_count = self.secondary_index.count_offsets()
        print(f"B+tree (secondary) keys: {bplus_key_count}")
        print(f"B+tree (secondary) offsets: {bplus_offset_count}")

        # Validate B+tree structure
        validation = self.secondary_index.validate()
        if not validation['valid']:
            print(f"WARNING: B+tree validation errors: {validation['errors'][:3]}")
            print(f"         Total errors: {len(validation['errors'])}")

        category_stats = defaultdict(int)
        price_stats = {'min': float('inf'), 'max': float('-inf'), 'total': 0.0}

        for product in all_products:
            category_stats[product.category] += 1
            price_stats['min'] = min(price_stats['min'], product.price)
            price_stats['max'] = max(price_stats['max'], product.price)
            price_stats['total'] += product.price

        print("\nProducts by category:")
        for category, count in sorted(category_stats.items()):
            print(f"  {category}: {count}")

        if all_products:
            print(f"\nPrice statistics:")
            print(f"  Min price: ${price_stats['min']:.2f}")
            print(f"  Max price: ${price_stats['max']:.2f}")
            print(f"  Avg price: ${price_stats['total'] / len(all_products):.2f}")

        if os.path.exists(self.data_file):
            file_size = os.path.getsize(self.data_file)
            print(f"\nFile size: {file_size:,} bytes")
            total_records = len(all_products) + len(self.deleted_offsets)
            utilization = (len(all_products) * 100) / max(1, total_records)
            print(f"File utilization: {utilization:.1f}%")

        print("=" * 80)


# ========================
# 5. TEST FUNCTION
# ========================
def run_complete_test():
    print("=" * 80)
    print("E-COMMERCE DATABASE WITH MULTI-LEVEL INDEX - COMPLETE TEST")
    print("=" * 80)

    if os.path.exists("products.dat"):
        os.remove("products.dat")

    print("\n1. INITIALIZING DATABASE...")
    db = ProductDatabase("products.dat")

    products_data = [
        (1001, "Wireless Mouse", "Electronics", 25.99, 150),
        (1002, "Bluetooth Headphones", "Electronics", 79.99, 85),
        (1003, "Coffee Maker", "Home Appliances", 39.95, 45),
        (1004, "Yoga Mat", "Sports", 18.50, 200),
        (1005, "Digital Camera", "Photography", 299.99, 30),
        (1006, "Electric Toothbrush", "Personal Care", 49.99, 120),
        (1007, "LED Desk Lamp", "Office Supplies", 23.99, 60),
        (1008, "Gaming Keyboard", "Electronics", 89.99, 40),
        (1009, "Fitness Tracker", "Wearables", 129.99, 75),
        (1010, "Smartphone Case", "Accessories", 12.99, 350),
        (1011, "Water Bottle", "Sports", 15.00, 250),
        (1012, "Noise Cancelling Earbuds", "Electronics", 199.99, 50),
        (1013, "Blender", "Home Appliances", 99.95, 65),
        (1014, "Running Shoes", "Footwear", 120.00, 80),
        (1015, "Laptop Stand", "Office Supplies", 27.99, 90),
        (1016, "Electric Kettle", "Home Appliances", 29.95, 110),
        (1017, "Bluetooth Speaker", "Electronics", 59.99, 70),
        (1018, "Hair Dryer", "Personal Care", 39.99, 55),
        (1019, "Action Camera", "Photography", 199.99, 25),
        (1020, "Smartwatch", "Wearables", 249.99, 40)
    ]

    print("\n2. INSERTING 20 PRODUCTS...")
    for data in products_data:
        db.insert(*data)

    db.statistics()

    print("\n" + "=" * 80)
    print("TESTING QUERIES")
    print("=" * 80)

    print("\n3. Search by ID:")
    print("   SELECT * WHERE ID=1005")
    results = db.select("ID=1005")
    print(f"   Found {len(results)} product(s):")
    for product in results:
        print(f"   {product}")

    print("\n4. Search by category:")
    print("   SELECT * WHERE CATEGORY='Electronics'")
    results = db.select("CATEGORY='Electronics'")
    print(f"   Found {len(results)} Electronics products:")
    for product in results:
        print(f"   {product}")

    print("\n5. Range query:")
    print("   SELECT * WHERE CATEGORY='Electronics' AND PRICE BETWEEN 50 AND 100")
    results = db.select("CATEGORY='Electronics' AND PRICE BETWEEN 50 AND 100")
    print(f"   Found {len(results)} products:")
    for product in results:
        print(f"   {product}")

    print("\n6. OR condition:")
    print("   SELECT * WHERE ID=1010 OR CATEGORY='Sports'")
    results = db.select("ID=1010 OR CATEGORY='Sports'")
    print(f"   Found {len(results)} products:")
    for product in results:
        print(f"   {product}")

    print("\n7. Price range query:")
    print("   SELECT * WHERE PRICE BETWEEN 20 AND 50")
    results = db.select("PRICE BETWEEN 20 AND 50")
    print(f"   Found {len(results)} products:")
    for product in results:
        print(f"   {product}")

    print("\n" + "=" * 80)
    print("TESTING DELETE OPERATIONS")
    print("=" * 80)

    print("\n8. Delete by ID:")
    print("   DELETE WHERE ID=1005")
    deleted = db.delete("ID=1005")
    print(f"   Deleted {deleted} product(s)")

    print("\n   Verify deletion - search for ID=1005:")
    results = db.select("ID=1005")
    if results:
        print("   ERROR: Product still found!")
    else:
        print("   SUCCESS: Product not found")

    print("\n9. Delete by category and price:")
    print("   DELETE WHERE CATEGORY='Electronics' AND PRICE BETWEEN 70 AND 90")
    deleted = db.delete("CATEGORY='Electronics' AND PRICE BETWEEN 70 AND 90")
    print(f"   Deleted {deleted} product(s)")

    print("\n   Remaining Electronics products:")
    results = db.select("CATEGORY='Electronics'")
    for product in results:
        print(f"   {product}")

    print("\n" + "=" * 80)
    print("UPDATED STATISTICS")
    print("=" * 80)
    db.statistics()

    print("\n" + "=" * 80)
    print("TESTING INSERT AFTER DELETION")
    print("=" * 80)

    print("\n10. Insert new product with ID=1021:")
    db.insert(1021, "Tablet", "Electronics", 399.99, 25)
    db.insert(1022, "Laptop", "Electronics", 999.99, 50)

    print("\n11. Compact database...")
    db.compact()

    print("\n" + "=" * 80)
    print("FINAL TESTS AFTER COMPACTION")
    print("=" * 80)

    print("\n12. Search for all Electronics after compaction:")
    results = db.select("CATEGORY='Electronics'")
    print(f"    Found {len(results)} Electronics products:")
    for product in results:
        print(f"    {product}")

    print("\n13. Complex OR query:")
    print("    SELECT * WHERE CATEGORY='Home Appliances' OR PRICE BETWEEN 20 AND 40")
    results = db.select("CATEGORY='Home Appliances' OR PRICE BETWEEN 20 AND 40")
    print(f"    Found {len(results)} products:")
    for product in sorted(results, key=lambda x: (x.category, x.price)):
        print(f"    {product}")

    print("\n14. ID range query:")
    print("    SELECT * WHERE ID BETWEEN 1010 AND 1015")
    results = db.select("ID BETWEEN 1010 AND 1015")
    print(f"    Found {len(results)} products:")
    for product in results:
        print(f"    {product}")

    print("\n" + "=" * 80)
    print("FINAL STATISTICS")
    print("=" * 80)
    db.statistics()

    print("\n" + "=" * 80)
    print("TEST COMPLETED SUCCESSFULLY!")
    print("=" * 80)


# ========================
# 6. MAIN ENTRY POINT
# ========================
if __name__ == "__main__":
    run_complete_test()
````
